```python
import torch

embedding = torch.randn(100, 4)
tokens = torch.tensor([
    [4, 5, 6],
    [2, 3, 4],
    [6, 7, 8],
    [8, 9, 10]
])

x = embedding[tokens]
print(x)
print(x.shape)
```

```python
print(x[0])
print(x[0].shape)
```

```python
# select first sentence, first token => 4 numbers
print(x[0, 0])
print(x[0, 0].shape)
```

```python
# select 1st token from all the sentences
print(x[:, 0, :])
print(x[:, 0, :].shape)
```

```python
print(x[:, :, 0])
print(x[:, :, 0].shape)
```

```python
print(x[0, :, :])
print(x[0, :, :].shape)
```
