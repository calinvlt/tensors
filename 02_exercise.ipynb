```python
import torch

vocab_size = 10_000
embedding_dim = 8

# generate random embeddings for all the vocabulary
embedding = torch.randn(vocab_size, embedding_dim)
print("Embedding shape", embedding.shape)

# 4 sentences, 3 tokens each
tokens = torch.tensor([
    [12, 57, 83],
    [12, 57, 91],
    [12, 64, 72],
    [12, 88, 93]
])
print("Tokens shape", tokens.shape)

# Lookup embeddings
x = embedding[tokens]
print("Output shape", x.shape)
```
