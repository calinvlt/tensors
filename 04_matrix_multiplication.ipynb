# Matrix multiplication

```python
import torch

x = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])

W = torch.tensor([
    [1., 2.],
    [3., 4.],
    [5., 6.],
])

y = x @ W

print(y)
print(x.shape, W.shape, y.shape)
```

## Shape Rule
(2, 3) @ (3, 2) -> (2, 2)
(A, B) @ (B, C) -> (A, C)

When: Q = x @ W_q
x (batch, tokens, embedding) @ W_q -> Q (batch, token, head_dimension)

## Transformer

x: (batch, token, dimensions) @ W_q: (batch, token, head_dimensions) -> Q: (batch, token, head_dimensions)

x: (batch, token, dimensions) @ W_k: (batch, token, head_dimensions) -> K: (batch, token, head_dimensions)

### For a transformer head
Q: (batch, token, head_dimensions) @ KT (batch, head_dimensions, token) -> attention_scores: (batch, token, token)
