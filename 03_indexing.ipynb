# PyTorch Indexing

```python
import torch

embedding = torch.tensor([
    [10, 11, 12],
    [20, 21, 22],
    [30, 31, 32],
    [40, 41, 42],
])
embedding.shape
```

token 0 -> [10, 11, 12]

token 1 -> [20, 21, 22]

token 2 -> [30, 31, 32]

token 3 -> [40, 41, 42]

```python
tokens = torch.tensor([2, 0, 3])
embedding[tokens]
```

Embedding lookup is basically: **use each token ID as a row number in the embedding matrix**
