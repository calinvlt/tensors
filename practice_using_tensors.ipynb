Tensor = a block of numbers with a shape, where each dimension has a meaning.
In an LLM, the dimensions have semantic meaning.
`(batch, sequence, embedding)`
`(4, 10, 768)`
- 4 = 4 sentences
- 10 = 10 tokens each sentence
- 768 = each token represented by 768
This type of tensor is the basic object that flows through the Transformer.

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
x.shape
```

```python
x = torch.tensor([1, 2, 3, 4])
print(x)
print(x.shape)
print(x.dtype)
```

```python
# Vector
x = torch.tensor([1, 2, 3])
x.shape
```

```python
# 2D or a matrix
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
x.shape
```

I  like cats
|    |   |
12  57   83 
Each word correspond with the number of the word from the dictionary associated with all the words in the vocabulary.

```python
tokens = torch.tensor([12, 57, 83])
tokens.shape
```

```python
# embedding matrix for the entire vocabulary (10k tokens each having 768 tokens) => matrix with 10K rows and 768 columns
embeddings = torch.randn(10_000, 768)
embeddings.shape

# every token is identified by the row number and each token is represented by 768 numbers which are the number of the columns
```

```python
# Getting the embedding for the sentence
x = embeddings[tokens]
x.shape
# from the embedding matrix only rows 12, 57 and 83 were selected; they corespond with the numbers from the 'tokens' variable
```

## Mental model
"I"    -> [768 numbers]
"like" -> [768 numbers]
"cats" -> [768 numbers]

3 tokens X 768 numbers/token

```python
## Adding a batch. 5 sentences each containing 3 words(tokens)
tokens = torch.tensor([
    [12, 57, 83],
    [12, 57, 91],
    [12, 64, 72],
    [12, 88, 93],
    [12, 64, 93],
])
tokens.shape

x = embeddings[tokens]
x.shape
# this is the fundamental LLM tensor representation (batch, tokens, embeddings)
```

## LLM Mental Mode
(5, 3, 768)
- 5 sentences
- each sentence has 3 tokens
- each token has or is represented by 768 embeddings
